## 0. Előkészítés – az adatfájlok letöltése

Futtasd ezt a cellát: letölti a feladathoz tartozó fájlokat (`students.csv`, `grades.csv`) a GitHub-repóból a `case_study_data/` mappába. A feladatokban innen olvasd be őket, pl. `open('case_study_data/students.csv')`.

```python
import os
import urllib.request

BASE = 'https://raw.githubusercontent.com/domahet/python-exercise/main/02_diak_jegykezelo/data'
FILES = ['students.csv', 'grades.csv']
DATA_DIR = 'case_study_data'

os.makedirs(DATA_DIR, exist_ok=True)
for name in FILES:
    path = os.path.join(DATA_DIR, name)
    if not os.path.exists(path):
        urllib.request.urlretrieve(f'{BASE}/{name}', path)

print('Adatfájlok:', sorted(os.listdir(DATA_DIR)))
```

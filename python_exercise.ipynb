# Python exercise – Céges autók

Egy cég **10 autóval** rendelkezik, amelyeket a dolgozók üzleti ügyeik intézésére igénybe vehetnek. Egy autót akár többnapos útra is elvihetnek, és egy autót egy nap több dolgozó is elvihet. A rendszer az autók parkolóból való **ki- és behajtását** rögzíti.

- A parkoló minden nap **7–23 óra** között van nyitva, csak ekkor lehet autót elvinni és visszahozni.
- Az autót mindig az a dolgozó hozza vissza, aki elvitte. Egyszerre egy dolgozónál legfeljebb egy autó lehet.
- A hónap első napján mind a tíz autó a parkolóban volt.

## A forrásfájl: `autok.txt`

Egy hónap (30 nap) adatai. Soronként 6, szóközzel elválasztott adat (legfeljebb 500 sor), napok, azon belül időpont szerint rendezve:

| adat | típus | jelentés |
|---|---|---|
| nap | egész (1–30) | a hónap adott napja |
| óra:perc | szöveg (`óó:pp`) | a ki- vagy behajtás időpontja |
| rendszám | 6 karakteres szöveg (`CEG300`–`CEG309`) | az autó rendszáma |
| személy azonosítója | egész (500–600) | az autót igénybe vevő dolgozó |
| km számláló | egész | a km-számláló állása |
| ki/be hajtás | `0` vagy `1` | **0** = kihajtás, **1** = behajtás |

Példa:
```
5   07:30  CEG300  590   30580  0
5   14:16  CEG300  590   30656  1
5   17:00  CEG300  534   30656  0
5   19:03  CEG300  534   30784  1
```
A CEG300-as autót az 5. napon 7:30-kor vitte el az 590-es dolgozó (30 580 km) és 14:16-kor hozta vissza (30 656 km); 17:00-kor az 534-es dolgozó vitte el, 19:03-kor hozta vissza.

**Szabályok:** a bemenő adatok helyességét nem kell ellenőrizni. A képernyőre írást igénylő feladatok előtt írd ki a feladat sorszámát (pl. `3. feladat`). Ha adatot kérsz be, jelenítsd meg, milyen értéket vársz.
